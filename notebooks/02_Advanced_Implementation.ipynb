```python
# Load the original training datasets

train_transaction = pd.read_csv(
    os.path.join(RAW_DIR, "train_transaction.csv")
)

train_identity = pd.read_csv(
    os.path.join(RAW_DIR, "train_identity.csv")
)

print("Transaction shape:", train_transaction.shape)
print("Identity shape:", train_identity.shape)

# Merge datasets exactly as Tuesday

train = train_transaction.merge(
    train_identity,
    on="TransactionID",
    how="left"
)

print("Merged dataset:", train.shape)

# Recreate Tuesday's chronological split

train_sorted = train.sort_values(
    by="TransactionDT"
).reset_index(drop=True)

total_rows = len(train_sorted)

train_end = int(total_rows * 0.70)
val_end = int(total_rows * 0.85)

train_set = train_sorted.iloc[:train_end].copy()
val_set = train_sorted.iloc[train_end:val_end].copy()
test_set = train_sorted.iloc[val_end:].copy()

print("\nChronological split completed!")

print("Training:", train_set.shape)
print("Validation:", val_set.shape)
print("Testing:", test_set.shape)

print("\nTime boundaries:")

for name, dataset in [
    ("Training", train_set),
    ("Validation", val_set),
    ("Testing", test_set)
]:
    print(
        name,
        dataset["TransactionDT"].min(),
        "to",
        dataset["TransactionDT"].max()
    )
```

```python

# Separate features and target variables

X_train = train_set.drop(
    columns=["isFraud", "TransactionID"]
).copy()
y_train = train_set["isFraud"].copy()

X_val = val_set.drop(
    columns=["isFraud", "TransactionID"]
).copy()
y_val = val_set["isFraud"].copy()

X_test = test_set.drop(
    columns=["isFraud", "TransactionID"]
).copy()
y_test = test_set["isFraud"].copy()


# Identify categorical columns using training data
categorical_columns = X_train.select_dtypes(
    include=["object", "category"]
).columns.tolist()


# Apply training categories to validation and test data
for col in categorical_columns:
    X_train[col] = X_train[col].astype("category")

    categories = X_train[col].cat.categories

    X_val[col] = pd.Categorical(
        X_val[col],
        categories=categories
    )

    X_test[col] = pd.Categorical(
        X_test[col],
        categories=categories
    )


# Verify the prepared datasets
print("Feature preparation completed!\n")

print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("X_test:", X_test.shape)

print("\nTarget dimensions:")
print("y_train:", y_train.shape)
print("y_val:", y_val.shape)
print("y_test:", y_test.shape)

print("\nCategorical columns:", len(categorical_columns))
print("Fraud rate in training:", y_train.mean())
print("Fraud rate in validation:", y_val.mean())
```

```python

# Create balanced samples from training and validation periods

SAMPLE_SIZE = 50000

train_sample = X_train.sample(
    n=SAMPLE_SIZE,
    random_state=42
).drop(columns=["TransactionDT"])

val_sample = X_val.sample(
    n=SAMPLE_SIZE,
    random_state=42
).drop(columns=["TransactionDT"])


# Assign labels:
# 0 = Training period
# 1 = Validation period

train_sample["is_validation"] = 0
val_sample["is_validation"] = 1


# Combine both samples
adversarial_data = pd.concat(
    [train_sample, val_sample],
    ignore_index=True
)

# Separate features and adversarial target
X_adv = adversarial_data.drop(
    columns=["is_validation"]
)

y_adv = adversarial_data["is_validation"]


print("Adversarial dataset created!\n")
print("Dataset shape:", adversarial_data.shape)
print("Feature shape:", X_adv.shape)

print("\nClass distribution:")
print(y_adv.value_counts())

print("\nClass proportions:")
print(y_adv.value_counts(normalize=True))
```

```python

import numpy as np
import pandas as pd

# Top 20 features identified by adversarial validation
top_features = adv_importance["Feature"].head(20).tolist()

# Separate numerical and categorical features
categorical_top_features = [
    col for col in top_features
    if col in categorical_columns
    or col in ["card1", "card2", "card5", "addr1"]
]

numerical_top_features = [
    col for col in top_features
    if col not in categorical_top_features
]


def calculate_numeric_psi(reference, current, bins=10):
    reference = pd.to_numeric(reference, errors="coerce")
    current = pd.to_numeric(current, errors="coerce")

    # Build bins using the reference (training) distribution
    _, bin_edges = pd.qcut(
        reference.dropna(),
        q=bins,
        retbins=True,
        duplicates="drop"
    )

    bin_edges[0] = -np.inf
    bin_edges[-1] = np.inf

    ref_binned = pd.cut(reference, bins=bin_edges)
    cur_binned = pd.cut(current, bins=bin_edges)

    # Include missing values as a separate bin
    ref_counts = ref_binned.value_counts(sort=False).astype(float)
    cur_counts = cur_binned.value_counts(sort=False).astype(float)

    ref_counts.loc["MISSING"] = reference.isna().sum()
    cur_counts.loc["MISSING"] = current.isna().sum()

    ref_pct = ref_counts / len(reference)
    cur_pct = cur_counts / len(current)

    epsilon = 1e-6
    ref_pct = ref_pct.clip(lower=epsilon)
    cur_pct = cur_pct.clip(lower=epsilon)

    return np.sum(
        (cur_pct - ref_pct) * np.log(cur_pct / ref_pct)
    )


def calculate_categorical_psi(reference, current, top_n=50):
    # Use the most frequent training categories
    common_categories = (
        reference.value_counts(dropna=True)
        .head(top_n)
        .index
    )

    def group_categories(series):
        grouped = series.astype("object").copy()
        grouped = grouped.where(
            grouped.isin(common_categories),
            "OTHER"
        )
        grouped = grouped.where(
            series.notna(),
            "MISSING"
        )
        return grouped

    ref_grouped = group_categories(reference)
    cur_grouped = group_categories(current)

    categories = set(ref_grouped.unique()) | set(cur_grouped.unique())

    ref_pct = ref_grouped.value_counts(normalize=True).reindex(
        categories, fill_value=0
    )
    cur_pct = cur_grouped.value_counts(normalize=True).reindex(
        categories, fill_value=0
    )

    epsilon = 1e-6
    ref_pct = ref_pct.clip(lower=epsilon)
    cur_pct = cur_pct.clip(lower=epsilon)

    return np.sum(
        (cur_pct - ref_pct) * np.log(cur_pct / ref_pct)
    )


# Calculate PSI for all top 20 features
psi_results = []

for feature in top_features:
    if feature in categorical_top_features:
        psi = calculate_categorical_psi(
            X_train[feature],
            X_val[feature]
        )
        feature_type = "Categorical"
    else:
        psi = calculate_numeric_psi(
            X_train[feature],
            X_val[feature]
        )
        feature_type = "Numerical"

    psi_results.append({
        "Feature": feature,
        "Type": feature_type,
        "PSI": psi
    })


psi_comparison = pd.DataFrame(psi_results).sort_values(
    by="PSI",
    ascending=False
).reset_index(drop=True)

print("Training vs Validation – PSI Results\n")
print(psi_comparison.to_string(index=False))

# Save results
psi_comparison.to_csv(
    os.path.join(RESULTS_DIR, "adversarial_top20_psi.csv"),
    index=False
)

print("\nPSI results saved successfully!")
```

```python

# Create time-based features using the original TransactionDT

def create_time_features(dataset):
    transaction_dt = dataset["TransactionDT"]

    # Elapsed day index
    features = pd.DataFrame(index=dataset.index)

    features["transaction_day"] = (
        transaction_dt // 86400
    ).astype("int32")

    # Hour of day (0–23)
    transaction_hour = (
        (transaction_dt % 86400) // 3600
    ).astype("int8")

    features["transaction_hour"] = transaction_hour

    # Cyclical encoding of hour
    features["transaction_hour_sin"] = np.sin(
        2 * np.pi * transaction_hour / 24
    )

    features["transaction_hour_cos"] = np.cos(
        2 * np.pi * transaction_hour / 24
    )

    return features


# Generate features separately for each chronological split
time_train = create_time_features(train_set)
time_val = create_time_features(val_set)
time_test = create_time_features(test_set)


# Add them to the existing feature datasets
X_train = pd.concat([X_train, time_train], axis=1)
X_val = pd.concat([X_val, time_val], axis=1)
X_test = pd.concat([X_test, time_test], axis=1)


# Verify
print("Time-based features created!\n")

print("New features:")
print(time_train.head())

print("\nUpdated dimensions:")
print("Training:", X_train.shape)
print("Validation:", X_val.shape)
print("Testing:", X_test.shape)

print("\nHour range:")
print("Training:", time_train["transaction_hour"].min(),
      "to", time_train["transaction_hour"].max())
print("Validation:", time_val["transaction_hour"].min(),
      "to", time_val["transaction_hour"].max())
print("Testing:", time_test["transaction_hour"].min(),
      "to", time_test["transaction_hour"].max())
```

```python

import numpy as np

ALPHA = 0.25
GAMMA = 2.0


def focal_loss_gradient_hessian(y_true, raw_predictions,
                                alpha=ALPHA, gamma=GAMMA):

    # Convert raw predictions into probabilities
    p = 1.0 / (1.0 + np.exp(-raw_predictions))

    # Prevent log(0) and division by zero
    p = np.clip(p, 1e-12, 1.0 - 1e-12)

    # Class-specific parameters
    alpha_t = np.where(y_true == 1, alpha, 1.0 - alpha)

    # Correct-class probability
    p_t = np.where(y_true == 1, p, 1.0 - p)

    # Gradient and Hessian for positive and negative classes
    # Calculate positive-class formulas first
    pos_grad = (
        alpha * (1 - p) ** gamma
        * (gamma * p * np.log(p) - (1 - p))
    )

    pos_hess = (
        alpha * p * (1 - p) ** gamma
        * (
            (1 - p) * (2 * gamma + 1)
            - gamma * np.log(p) * (1 - (gamma + 1) * p)
        )
    )

    # For negative examples, use the symmetry:
    # FL_negative(p) = FL_positive(1-p), with alpha replaced by 1-alpha
    neg_grad = -(
        (1 - alpha) * p ** gamma
        * (gamma * (1 - p) * np.log(1 - p) - p)
    )

    neg_hess = (
        (1 - alpha) * (1 - p) * p ** gamma
        * (
            p * (2 * gamma + 1)
            - gamma * np.log(1 - p) * (1 - (gamma + 1) * (1 - p))
        )
    )

    grad = np.where(y_true == 1, pos_grad, neg_grad)
    hess = np.where(y_true == 1, pos_hess, neg_hess)

    return grad, hess


# Test with sample examples
sample_y = np.array([1, 1, 0, 0])
sample_logits = np.array([2.0, -1.0, -2.0, 1.0])

gradients, hessians = focal_loss_gradient_hessian(
    sample_y,
    sample_logits
)

print("Sample labels:", sample_y)
print("Sample raw predictions:", sample_logits)
print("Gradients:", gradients)
print("Hessians:", hessians)

print("\nHessian statistics:")
print("Minimum:", hessians.min())
print("Maximum:", hessians.max())
print("Negative Hessians:", np.sum(hessians < 0))
```

```python

def focal_loss_gradient_hessian(
    y_true,
    raw_predictions,
    alpha=ALPHA,
    gamma=GAMMA
):
    # Convert logits to probabilities
    p = 1.0 / (1.0 + np.exp(-raw_predictions))
    p = np.clip(p, 1e-12, 1.0 - 1e-12)

    # Positive-class gradient
    pos_grad = (
        alpha * (1 - p) ** gamma
        * (gamma * p * np.log(p) - (1 - p))
    )

    # Corrected positive-class Hessian
    pos_hess = (
        alpha * p * (1 - p) ** gamma
        * (
            (1 - p) * (2 * gamma + 1)
            + gamma * np.log(p)
            * (1 - (gamma + 1) * p)
        )
    )

    # Negative-class gradient
    neg_grad = -(
        (1 - alpha) * p ** gamma
        * (
            gamma * (1 - p) * np.log(1 - p) - p
        )
    )

    # Corrected negative-class Hessian
    neg_hess = (
        (1 - alpha) * (1 - p) * p ** gamma
        * (
            p * (2 * gamma + 1)
            + gamma * np.log(1 - p)
            * (1 - (gamma + 1) * (1 - p))
        )
    )

    grad = np.where(y_true == 1, pos_grad, neg_grad)
    hess = np.where(y_true == 1, pos_hess, neg_hess)

    return grad, hess


# Recheck sample examples
gradients, hessians = focal_loss_gradient_hessian(
    sample_y,
    sample_logits
)

print("Corrected gradients:", gradients)
print("Corrected Hessians:", hessians)

print("\nHessian statistics:")
print("Minimum:", hessians.min())
print("Maximum:", hessians.max())
print("Negative Hessians:", np.sum(hessians < 0))
```

```python

def focal_loss_from_logits(y_true, logits, alpha=ALPHA, gamma=GAMMA):
    p = 1.0 / (1.0 + np.exp(-logits))
    p = np.clip(p, 1e-12, 1.0 - 1e-12)

    p_t = np.where(y_true == 1, p, 1 - p)
    alpha_t = np.where(y_true == 1, alpha, 1 - alpha)

    loss = -alpha_t * (1 - p_t) ** gamma * np.log(p_t)

    return loss
```

```python

# Numerical verification using central finite differences

epsilon = 1e-5

numerical_gradients = []
numerical_hessians = []

for y, z in zip(sample_y, sample_logits):

    # Numerical gradient from the loss
    loss_plus = focal_loss_from_logits(
        np.array([y]), np.array([z + epsilon])
    )[0]

    loss_minus = focal_loss_from_logits(
        np.array([y]), np.array([z - epsilon])
    )[0]

    numerical_gradient = (
        loss_plus - loss_minus
    ) / (2 * epsilon)

    # Numerical Hessian from the analytical gradient
    grad_plus, _ = focal_loss_gradient_hessian(
        np.array([y]), np.array([z + epsilon])
    )

    grad_minus, _ = focal_loss_gradient_hessian(
        np.array([y]), np.array([z - epsilon])
    )

    numerical_hessian = (
        grad_plus[0] - grad_minus[0]
    ) / (2 * epsilon)

    numerical_gradients.append(numerical_gradient)
    numerical_hessians.append(numerical_hessian)


# Compare analytical and numerical results
analytical_gradients, analytical_hessians = (
    focal_loss_gradient_hessian(sample_y, sample_logits)
)

verification = pd.DataFrame({
    "Label": sample_y,
    "Logit": sample_logits,
    "Analytical Gradient": analytical_gradients,
    "Numerical Gradient": numerical_gradients,
    "Analytical Hessian": analytical_hessians,
    "Numerical Hessian": numerical_hessians
})

print(verification.to_string(index=False))

print("\nMaximum absolute gradient error:",
      np.max(np.abs(
          analytical_gradients - numerical_gradients
      )))

print("Maximum absolute Hessian error:",
      np.max(np.abs(
          analytical_hessians - numerical_hessians
      )))
```

```python

# LightGBM-compatible safeguarded Focal Loss objective

HESSIAN_FLOOR = 1e-6

def focal_loss_objective(y_true, raw_predictions):
    gradients, hessians = focal_loss_gradient_hessian(
        y_true,
        raw_predictions,
        alpha=ALPHA,
        gamma=GAMMA
    )

    # Keep exact gradients; safeguard negative/small Hessians
    safe_hessians = np.maximum(
        hessians,
        HESSIAN_FLOOR
    )

    return gradients, safe_hessians


# Test the safeguarded objective
safe_grad, safe_hess = focal_loss_objective(
    test_labels,
    test_logits
)

print("All gradients finite:",
      np.isfinite(safe_grad).all())

print("All Hessians finite:",
      np.isfinite(safe_hess).all())

print("Minimum safeguarded Hessian:",
      safe_hess.min())

print("Negative safeguarded Hessians:",
      np.sum(safe_hess < 0))

print("Hessians clipped:",
      np.sum(stress_hess < HESSIAN_FLOOR))
```

```python
# Select thresholds using calibration evaluation data

from sklearn.metrics import precision_score, recall_score, f1_score
from scipy.special import expit

validation_probabilities = {}

# BCE probabilities
validation_probabilities["BCE"] = expit(
    validation_raw_scores["BCE"][calibration_split:]
)

# Focal Alpha 0.75 calibrated probabilities
focal_platt = honest_calibrators["Focal Alpha 0.75"]["platt"]

validation_probabilities["Focal Alpha 0.75"] = (
    focal_platt.predict_proba(
        validation_raw_scores["Focal Alpha 0.75"][
            calibration_split:
        ].reshape(-1, 1)
    )[:, 1]
)

# Search thresholds
thresholds = np.arange(0.05, 0.951, 0.01)

final_thresholds = {}
threshold_selection_results = []

for model_name, probabilities in validation_probabilities.items():

    best_f1 = -1
    best_result = None

    for threshold in thresholds:

        predictions = (probabilities >= threshold).astype(int)

        precision = precision_score(
            y_cal_eval, predictions, zero_division=0
        )
        recall = recall_score(
            y_cal_eval, predictions, zero_division=0
        )
        f1 = f1_score(
            y_cal_eval, predictions, zero_division=0
        )

        if f1 > best_f1:

            best_f1 = f1

            best_result = {
                "Model": model_name,
                "Threshold": round(threshold, 2),
                "Precision": precision,
                "Recall": recall,
                "F1 Score": f1
            }

    final_thresholds[model_name] = best_result["Threshold"]
    threshold_selection_results.append(best_result)

threshold_selection_df = pd.DataFrame(
    threshold_selection_results
)

print("FINAL VALIDATION THRESHOLD SELECTION")
print("-" * 55)
print(threshold_selection_df.to_string(index=False))

print("\nFrozen thresholds:")
print(final_thresholds)
```

```python
# Categorical PSI across five chronological periods

categorical_drift_features = [
    "id_31",
    "card2",
    "card5"
]

def calculate_categorical_psi(reference, comparison):

    # Convert values to consistent strings
    reference = reference.astype("string").fillna("MISSING")
    comparison = comparison.astype("string").fillna("MISSING")

    # Combine all categories
    all_categories = sorted(
        set(reference.unique()) | set(comparison.unique())
    )

    reference_pct = (
        reference.value_counts(normalize=True)
        .reindex(all_categories, fill_value=0)
    )

    comparison_pct = (
        comparison.value_counts(normalize=True)
        .reindex(all_categories, fill_value=0)
    )

    epsilon = 1e-6

    reference_pct = reference_pct.clip(lower=epsilon)
    comparison_pct = comparison_pct.clip(lower=epsilon)

    psi = (
        (comparison_pct - reference_pct) *
        np.log(comparison_pct / reference_pct)
    ).sum()

    return psi


categorical_psi_results = []

for feature in categorical_drift_features:

    for period_name, period_data in test_periods.items():

        psi_value = calculate_categorical_psi(
            X_train[feature],
            period_data[feature]
        )

        categorical_psi_results.append({
            "Feature": feature,
            "Period": period_name,
            "PSI": psi_value
        })


categorical_psi_df = pd.DataFrame(categorical_psi_results)

print("CATEGORICAL FEATURE PSI")
print("=" * 65)

print(
    categorical_psi_df.pivot(
        index="Feature",
        columns="Period",
        values="PSI"
    ).round(4).to_string()
)
```

```python
# Conditional fraud-rate analysis for D11 and D15

def create_training_bins(reference, bins=5):

    reference = pd.to_numeric(reference, errors="coerce")

    edges = np.unique(
        reference.dropna().quantile(
            np.linspace(0, 1, bins + 1)
        ).values
    )

    if len(edges) < 2:
        raise ValueError("Not enough unique values to create bins.")

    edges[0] = -np.inf
    edges[-1] = np.inf

    return edges


def analyze_conditional_fraud_rate(feature):

    edges = create_training_bins(X_train[feature])

    datasets = {
        "Training": (X_train[feature], y_train)
    }

    for period_name, indices in zip(
        test_periods.keys(), test_indices
    ):
        datasets[period_name] = (
            X_test[feature].iloc[indices],
            np.asarray(y_test)[indices]
        )

    results = []

    for period_name, (feature_values, labels) in datasets.items():

        feature_values = pd.to_numeric(
            feature_values, errors="coerce"
        )

        bins_assigned = pd.cut(
            feature_values,
            bins=edges,
            include_lowest=True
        )

        bin_labels = bins_assigned.map(str).fillna("MISSING")

        analysis_data = pd.DataFrame({
            "Bin": bin_labels.to_numpy(),
            "isFraud": np.asarray(labels)
        })

        grouped = analysis_data.groupby(
            "Bin", observed=False
        )["isFraud"].agg(["count", "mean"])

        for bin_name, row in grouped.iterrows():

            results.append({
                "Feature": feature,
                "Period": period_name,
                "Bin": bin_name,
                "Transactions": int(row["count"]),
                "Fraud Rate": row["mean"]
            })

    return pd.DataFrame(results)


d11_conditional = analyze_conditional_fraud_rate("D11")
d15_conditional = analyze_conditional_fraud_rate("D15")

conditional_fraud_results = pd.concat(
    [d11_conditional, d15_conditional],
    ignore_index=True
)

print("CONDITIONAL FRAUD RATE ANALYSIS")
print("=" * 75)

for feature in ["D11", "D15"]:

    print(f"\nFEATURE: {feature}")

    display(
        conditional_fraud_results[
            conditional_fraud_results["Feature"] == feature
        ].pivot(
            index="Bin",
            columns="Period",
            values="Fraud Rate"
        ).round(4)
    )
```

```python
from scipy.stats import chi2_contingency

# Define the five groups we want to investigate
def selected_groups(X):

    return {
        "D11_MISSING": X["D11"].isna(),
        "D15_MISSING": X["D15"].isna(),
        "C13_BELOW_1": X["C13"] < 1,
        "id_13_BELOW_41": X["id_13"] < 41,
        "id_31_MISSING": X["id_31"].isna()
    }


# Training group masks and labels
train_masks = selected_groups(X_train)
train_labels = np.asarray(y_train)

# Prepare test period data
period_data = {}

for period_name, indices in zip(test_periods.keys(), test_indices):

    period_X = X_test.iloc[indices]
    period_y = np.asarray(y_test)[indices]

    period_data[period_name] = (
        selected_groups(period_X),
        period_y
    )


results = []

for group_name, train_mask in train_masks.items():

    train_mask = np.asarray(train_mask)
    train_group_labels = train_labels[train_mask]

    train_fraud = int(train_group_labels.sum())
    train_total = len(train_group_labels)
    train_nonfraud = train_total - train_fraud
    train_rate = train_fraud / train_total if train_total else np.nan

    for period_name, (period_masks, period_labels) in period_data.items():

        period_mask = np.asarray(period_masks[group_name])
        period_group_labels = period_labels[period_mask]

        period_fraud = int(period_group_labels.sum())
        period_total = len(period_group_labels)
        period_nonfraud = period_total - period_fraud
        period_rate = (
            period_fraud / period_total
            if period_total else np.nan
        )

        # 2x2 contingency table:
        # Rows = Training and Test period
        # Columns = Fraud and Non-fraud
        contingency = np.array([
            [train_fraud, train_nonfraud],
            [period_fraud, period_nonfraud]
        ])

        chi2, p_value, dof, expected = chi2_contingency(
            contingency,
            correction=False
        )

        results.append({
            "Group": group_name,
            "Period": period_name,
            "Training_Count": train_total,
            "Period_Count": period_total,
            "Training_Fraud_Rate": train_rate,
            "Period_Fraud_Rate": period_rate,
            "Difference_pp": (period_rate - train_rate) * 100,
            "Chi2_Statistic": chi2,
            "P_Value": p_value
        })


drift_tests = pd.DataFrame(results)

print("STATISTICAL TEST RESULTS")
print("=" * 90)

display(
    drift_tests[
        [
            "Group",
            "Period",
            "Training_Count",
            "Period_Count",
            "Training_Fraud_Rate",
            "Period_Fraud_Rate",
            "Difference_pp",
            "P_Value"
        ]
    ].round(5)
)
```

```python
# Benjamini-Hochberg False Discovery Rate correction

p_values = drift_tests["P_Value"].values

sorted_indices = np.argsort(p_values)
sorted_p = p_values[sorted_indices]

m = len(p_values)

adjusted_sorted = sorted_p * m / np.arange(1, m + 1)

# Enforce monotonicity
adjusted_sorted = np.minimum.accumulate(
    adjusted_sorted[::-1]
)[::-1]

adjusted_sorted = np.clip(adjusted_sorted, 0, 1)

adjusted_p = np.empty(m)
adjusted_p[sorted_indices] = adjusted_sorted

drift_tests["Adjusted_P_Value"] = adjusted_p

drift_tests["Significant"] = (
    drift_tests["Adjusted_P_Value"] < 0.05
)

print("MULTIPLE-TESTING CORRECTED RESULTS")
print("=" * 75)

display(
    drift_tests[
        [
            "Group",
            "Period",
            "Difference_pp",
            "P_Value",
            "Adjusted_P_Value",
            "Significant"
        ]
    ].round(5)
)

print(
    "\nSignificant comparisons:",
    int(drift_tests["Significant"].sum()),
    "out of",
    len(drift_tests)
)
```

```python
# Prepare grouped data for logistic interaction testing

interaction_data = []

# Training period
for group_name, mask in train_masks.items():

    mask = np.asarray(mask)
    labels = train_labels

    for group_value in [0, 1]:

        selected = labels[mask == bool(group_value)]

        interaction_data.append({
            "Group": group_name,
            "Period": "Training",
            "Group_Member": group_value,
            "Transactions": len(selected),
            "Frauds": int(selected.sum())
        })


# Five test periods
for group_name in train_masks.keys():

    for period_name, (period_masks, period_labels) in period_data.items():

        mask = np.asarray(period_masks[group_name])

        for group_value in [0, 1]:

            selected = period_labels[mask == bool(group_value)]

            interaction_data.append({
                "Group": group_name,
                "Period": period_name,
                "Group_Member": group_value,
                "Transactions": len(selected),
                "Frauds": int(selected.sum())
            })


interaction_df = pd.DataFrame(interaction_data)

interaction_df["NonFrauds"] = (
    interaction_df["Transactions"] - interaction_df["Frauds"]
)

interaction_df["Fraud_Rate"] = (
    interaction_df["Frauds"] / interaction_df["Transactions"]
)

period_order = [
    "Training", "Period 1", "Period 2",
    "Period 3", "Period 4", "Period 5"
]

interaction_df["Period"] = pd.Categorical(
    interaction_df["Period"],
    categories=period_order,
    ordered=True
)

print("Grouped interaction dataset:")
print("Rows:", len(interaction_df))
display(interaction_df.head(12))
```

```python
from scipy.stats import chi2

interaction_results = []

for group_name in interaction_df["Group"].unique():

    subset = interaction_df[
        interaction_df["Group"] == group_name
    ].copy()

    # Reduced model: period and group have separate effects,
    # but the group association remains constant across periods.
    reduced_model = smf.glm(
        formula="Fraud_Rate ~ C(Period) + Group_Member",
        data=subset,
        family=sm.families.Binomial(),
        freq_weights=subset["Transactions"]
    ).fit()

    # Full model: allow the group association to vary by period.
    full_model = smf.glm(
        formula="Fraud_Rate ~ C(Period) * Group_Member",
        data=subset,
        family=sm.families.Binomial(),
        freq_weights=subset["Transactions"]
    ).fit()

    # Likelihood-ratio test
    lr_statistic = reduced_model.deviance - full_model.deviance
    df_difference = (
        reduced_model.df_resid - full_model.df_resid
    )

    p_value = chi2.sf(lr_statistic, df_difference)

    interaction_results.append({
        "Group": group_name,
        "LR_Statistic": lr_statistic,
        "Degrees_of_Freedom": int(df_difference),
        "P_Value": p_value
    })


interaction_results = pd.DataFrame(interaction_results)

# Benjamini-Hochberg correction across the five group-level tests
p_values = interaction_results["P_Value"].values
m = len(p_values)

sorted_indices = np.argsort(p_values)
sorted_p = p_values[sorted_indices]

adjusted_sorted = sorted_p * m / np.arange(1, m + 1)
adjusted_sorted = np.minimum.accumulate(
    adjusted_sorted[::-1]
)[::-1]

adjusted_p = np.empty(m)
adjusted_p[sorted_indices] = np.clip(adjusted_sorted, 0, 1)

interaction_results["Adjusted_P_Value"] = adjusted_p
interaction_results["Significant"] = (
    interaction_results["Adjusted_P_Value"] < 0.05
)

print("GROUP × PERIOD INTERACTION TEST RESULTS")
print("=" * 75)

display(interaction_results.round(6))
```

```python
import polars as pl
import pandas as pd
import numpy as np
from collections import defaultdict
from pathlib import Path

RESULTS = Path(r"D:\Fraud-Detection\results")

# --------------------------------------------------
# 1. Load chronologically ordered training partition
# --------------------------------------------------
print("Loading training partition...")

train_history_df = (
    pl.scan_parquet(RESULTS / "reconstruction_pandas_order.parquet")
    .slice(0, 413378)
    .select(["card1", "addr1", "TransactionAmt"])
    .collect()
    .to_pandas()
)

print("Training rows:", len(train_history_df))

# --------------------------------------------------
# 2. Build historical state using training data
# --------------------------------------------------
card_count = defaultdict(int)
card_amount_sum = defaultdict(float)
card_amount_count = defaultdict(int)
addr_count = defaultdict(int)

print("\nBuilding training history...")

for card, addr, amount in zip(
    train_history_df["card1"].to_numpy(),
    train_history_df["addr1"].to_numpy(),
    train_history_df["TransactionAmt"].to_numpy()
):
    if pd.notna(card):
        card_count[card] += 1

        if pd.notna(amount):
            card_amount_sum[card] += float(amount)
            card_amount_count[card] += 1

    if pd.notna(addr):
        addr_count[addr] += 1

print("Unique card keys:", len(card_count))
print("Unique address keys:", len(addr_count))

# --------------------------------------------------
# 3. Calculate validation historical features
# --------------------------------------------------
print("\nCalculating validation historical features...")

card_prev_counts = []
card_prev_amount_means = []
addr_prev_counts = []

for card, addr, amount in zip(
    rebuilt_X_val["card1"].to_numpy(),
    rebuilt_X_val["addr1"].to_numpy(),
    rebuilt_X_val["TransactionAmt"].to_numpy()
):
    # Read history BEFORE updating it
    if pd.notna(card):
        previous_card_count = card_count[card]
        previous_amount_count = card_amount_count[card]

        previous_amount_mean = (
            card_amount_sum[card] / previous_amount_count
            if previous_amount_count > 0
            else np.nan
        )
    else:
        previous_card_count = 0
        previous_amount_mean = np.nan

    previous_addr_count = (
        addr_count[addr] if pd.notna(addr) else 0
    )

    card_prev_counts.append(previous_card_count)
    card_prev_amount_means.append(previous_amount_mean)
    addr_prev_counts.append(previous_addr_count)

    # Update history AFTER calculating current row
    if pd.notna(card):
        card_count[card] += 1

        if pd.notna(amount):
            card_amount_sum[card] += float(amount)
            card_amount_count[card] += 1

    if pd.notna(addr):
        addr_count[addr] += 1

# --------------------------------------------------
# 4. Attach the three historical features
# --------------------------------------------------
rebuilt_X_val["card1_prev_count"] = np.array(
    card_prev_counts, dtype="int64"
)

rebuilt_X_val["card1_prev_amount_mean"] = np.array(
    card_prev_amount_means, dtype="float64"
)

rebuilt_X_val["addr1_prev_count"] = np.array(
    addr_prev_counts, dtype="int64"
)

print("\nHistorical features created.")

for col in [
    "card1_prev_count",
    "card1_prev_amount_mean",
    "addr1_prev_count"
]:
    print(
        col,
        "| Missing:", rebuilt_X_val[col].isna().sum(),
        "| Min:", rebuilt_X_val[col].min(),
        "| Max:", rebuilt_X_val[col].max()
    )

print("\nFinal validation shape:", rebuilt_X_val.shape)
```

```python
import polars as pl
import pandas as pd
import numpy as np
import joblib
from pathlib import Path
from collections import defaultdict

PROJECT = Path(r"D:\Fraud-Detection")
RESULTS = PROJECT / "results"

# --------------------------------------------------
# 1. Load chronological training partition
# --------------------------------------------------
print("1. Loading training partition...")

train_df = (
    pl.scan_parquet(RESULTS / "reconstruction_pandas_order.parquet")
    .slice(0, 413378)
    .collect()
    .to_pandas()
)

print("Training rows:", len(train_df))

# --------------------------------------------------
# 2. Separate features and target
# --------------------------------------------------
print("\n2. Separating features and target...")

y_train_rebuilt = train_df["isFraud"].copy().reset_index(drop=True)

X_train_rebuilt = train_df.drop(
    columns=["isFraud", "TransactionID"]
).copy().reset_index(drop=True)

del train_df

# --------------------------------------------------
# 3. Restore categorical data types
# --------------------------------------------------
print("\n3. Restoring categorical columns...")

categorical_columns = X_train_rebuilt.select_dtypes(
    include=["object", "category", "str"]
).columns.tolist()

for col in categorical_columns:
    X_train_rebuilt[col] = X_train_rebuilt[col].astype("category")

print("Categorical columns:", len(categorical_columns))

# --------------------------------------------------
# 4. Missingness indicators
# --------------------------------------------------
print("\n4. Creating missingness indicators...")

X_train_rebuilt["id_31_is_missing"] = (
    X_train_rebuilt["id_31"].isna().astype("int8")
)

X_train_rebuilt["id_13_is_missing"] = (
    X_train_rebuilt["id_13"].isna().astype("int8")
)

# --------------------------------------------------
# 5. Time features
# --------------------------------------------------
print("\n5. Creating time features...")

dt = X_train_rebuilt["TransactionDT"]

X_train_rebuilt["transaction_day"] = (
    dt // 86400
).astype("int32")

X_train_rebuilt["transaction_hour"] = (
    (dt % 86400) // 3600
).astype("int8")

X_train_rebuilt["transaction_hour_sin"] = np.sin(
    2 * np.pi * X_train_rebuilt["transaction_hour"] / 24
)

X_train_rebuilt["transaction_hour_cos"] = np.cos(
    2 * np.pi * X_train_rebuilt["transaction_hour"] / 24
)

# --------------------------------------------------
# 6. Historical features
# --------------------------------------------------
print("\n6. Calculating historical features...")

card_count = defaultdict(int)
card_amount_sum = defaultdict(float)
card_amount_count = defaultdict(int)
addr_count = defaultdict(int)

card_prev_counts = []
card_prev_means = []
addr_prev_counts = []

cards = X_train_rebuilt["card1"].to_numpy()
addresses = X_train_rebuilt["addr1"].to_numpy()
amounts = X_train_rebuilt["TransactionAmt"].to_numpy()

for card, addr, amount in zip(cards, addresses, amounts):

    # Read previous history before updating
    card_prev_counts.append(
        card_count[card] if pd.notna(card) else 0
    )

    if pd.notna(card) and card_amount_count[card] > 0:
        card_prev_means.append(
            card_amount_sum[card] / card_amount_count[card]
        )
    else:
        card_prev_means.append(np.nan)

    addr_prev_counts.append(
        addr_count[addr] if pd.notna(addr) else 0
    )

    # Update history after calculating current row
    if pd.notna(card):
        card_count[card] += 1

        if pd.notna(amount):
            card_amount_sum[card] += float(amount)
            card_amount_count[card] += 1

    if pd.notna(addr):
        addr_count[addr] += 1

X_train_rebuilt["card1_prev_count"] = np.array(
    card_prev_counts, dtype="int64"
)

X_train_rebuilt["card1_prev_amount_mean"] = np.array(
    card_prev_means, dtype="float64"
)

X_train_rebuilt["addr1_prev_count"] = np.array(
    addr_prev_counts, dtype="int64"
)

# --------------------------------------------------
# 7. Final checks
# --------------------------------------------------
print("\n" + "=" * 55)
print("TRAINING RECONSTRUCTION SUMMARY")
print("=" * 55)

print("X_train shape:", X_train_rebuilt.shape)
print("y_train shape:", y_train_rebuilt.shape)
print("Categorical columns:", len(categorical_columns))
print("Fraud transactions:", int(y_train_rebuilt.sum()))
print("Fraud rate:", round(y_train_rebuilt.mean(), 6))
print("Historical card keys:", len(card_count))
print("Historical address keys:", len(addr_count))

print("\nExpected X_train: (413378, 441)")
print("Expected y_train: (413378,)")

print("\nReconstruction cell completed.")
```

```python
from pathlib import Path
import hashlib

RESULTS = Path(r"D:\Fraud-Detection\results")
BACKUP = RESULTS / "verified_reconstruction_backup"

files = [
    "X_train_reconstructed_verified.pkl",
    "y_train_reconstructed_verified.pkl",
    "X_val_reconstructed_verified.pkl",
    "y_val_reconstructed_verified.pkl"
]

def sha256(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(8 * 1024 * 1024), b""):
            h.update(chunk)
    return h.hexdigest()

for name in files:
    original_hash = sha256(RESULTS / name)
    backup_hash = sha256(BACKUP / name)

    print(name, "->", "VERIFIED" if original_hash == backup_hash else "MISMATCH")
```

```python
import lightgbm as lgb
import pandas as pd
import time
import gc

from sklearn.metrics import roc_auc_score, average_precision_score

# Select 439 features
features = experiments["no_missingness"]

X_train_exp = X_train_ablation[features]
X_val_exp = X_val_ablation[features]

print("Experiment: No Missingness Features")
print("Training shape:", X_train_exp.shape)
print("Validation shape:", X_val_exp.shape)

# Prepare LightGBM datasets
train_data = lgb.Dataset(
    X_train_exp,
    label=y_train_ablation,
    categorical_feature="auto"
)

val_data = lgb.Dataset(
    X_val_exp,
    label=y_val_ablation,
    categorical_feature="auto",
    reference=train_data
)

# Same BCE configuration as original experiment
bce_params = {
    "objective": "binary",
    "metric": "None",
    "boosting_type": "gbdt",
    "learning_rate": 0.03,
    "num_leaves": 31,
    "max_depth": -1,
    "min_data_in_leaf": 100,
    "min_sum_hessian_in_leaf": 0.01,
    "lambda_l1": 1.0,
    "lambda_l2": 10.0,
    "max_delta_step": 1.0,
    "feature_fraction": 0.8,
    "bagging_fraction": 0.8,
    "bagging_freq": 1,
    "verbosity": -1,
    "seed": 42,
    "num_threads": -1
}

def bce_evaluation(predictions, eval_data):
    from sklearn.metrics import roc_auc_score, average_precision_score
    import numpy as np

    y_true = eval_data.get_label()
    probabilities = np.clip(predictions, 1e-12, 1 - 1e-12)

    return [
        ("roc_auc", roc_auc_score(y_true, probabilities), True),
        ("pr_auc", average_precision_score(y_true, probabilities), True)
    ]

print("\nStarting training...")
start_time = time.time()

no_missingness_model = lgb.train(
    params=bce_params,
    train_set=train_data,
    num_boost_round=500,
    valid_sets=[train_data, val_data],
    valid_names=["training", "validation"],
    feval=bce_evaluation,
    callbacks=[
        lgb.early_stopping(stopping_rounds=50),
        lgb.log_evaluation(period=25)
    ]
)

training_time = time.time() - start_time

# Evaluate
predictions = no_missingness_model.predict(X_val_exp)

result = {
    "experiment": "no_missingness",
    "features": len(features),
    "trees": no_missingness_model.num_trees(),
    "roc_auc": roc_auc_score(y_val_ablation, predictions),
    "pr_auc": average_precision_score(y_val_ablation, predictions),
    "training_time_seconds": round(training_time, 2)
}

# Save model and result
no_missingness_model.save_model(
    str(MODELS / "ablation_no_missingness.txt")
)

pd.DataFrame([result]).to_csv(
    RESULTS / "ablation_no_missingness_metrics.csv",
    index=False
)

print("\nEXPERIMENT RESULT")
print(pd.DataFrame([result]).to_string(index=False))
print("\nModel and metrics saved.")

# Release temporary objects
del X_train_exp, X_val_exp, train_data, val_data, predictions
gc.collect()
```

```python
from scipy.special import expit
from sklearn.metrics import precision_score, recall_score, f1_score
import numpy as np
import pandas as pd

# Split validation data chronologically
calibration_split = len(y_val) // 2

y_cal_eval = np.asarray(y_val)[calibration_split:]

print("Calibration fitting rows:", calibration_split)
print("Threshold evaluation rows:", len(y_cal_eval))
print("Evaluation fraud rate:", y_cal_eval.mean())

# Generate validation probabilities
validation_probabilities = {}

# BCE: sigmoid conversion
validation_probabilities["BCE"] = expit(
    validation_raw_scores["BCE"][calibration_split:]
)

# Focal: saved Platt calibrator
validation_probabilities["Focal Alpha 0.75"] = (
    focal_platt.predict_proba(
        validation_raw_scores["Focal Alpha 0.75"][
            calibration_split:
        ].reshape(-1, 1)
    )[:, 1]
)

# Find the threshold with the highest validation F1
thresholds = np.arange(0.05, 0.951, 0.01)

final_thresholds = {}
threshold_selection_results = []

for model_name, probabilities in validation_probabilities.items():

    best_f1 = -1
    best_result = None

    for threshold in thresholds:

        predictions = (probabilities >= threshold).astype(int)

        precision = precision_score(
            y_cal_eval, predictions, zero_division=0
        )
        recall = recall_score(
            y_cal_eval, predictions, zero_division=0
        )
        f1 = f1_score(
            y_cal_eval, predictions, zero_division=0
        )

        if f1 > best_f1:
            best_f1 = f1

            best_result = {
                "Model": model_name,
                "Threshold": round(threshold, 2),
                "Precision": precision,
                "Recall": recall,
                "F1 Score": f1
            }

    final_thresholds[model_name] = best_result["Threshold"]
    threshold_selection_results.append(best_result)

threshold_selection_df = pd.DataFrame(threshold_selection_results)

print("\nFINAL VALIDATION THRESHOLDS")
display(threshold_selection_df.round(4))

print("\nThreshold dictionary:")
print(final_thresholds)
```
